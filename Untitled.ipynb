```python
import numpy as np
import matplotlib.pyplot as plt

# pの範囲
p = np.linspace(0.001, 0.999, 500)

# y=1 の場合の損失
loss_y1 = -np.log(p)
# y=0 の場合の損失
loss_y0 = -np.log(1-p)

plt.plot(p, loss_y1, label="y=1")
plt.plot(p, loss_y0, label="y=0")
plt.xlabel("予測確率 p")
plt.ylabel("損失 L")
plt.title("ロジスティック回帰の損失関数")
plt.legend()
plt.grid(True)
plt.show()
```
